## Data input

```python
import pandas as pd
url = "https://vincentarelbundock.github.io/Rdatasets/csv/survival/pbc.csv"
pbc = pd.read_csv(url)
```

## Non-trivial transformation/analysis

```python
print(pbc["platelet"].mean())
print(pbc["platelet"].median())
max_value = (pbc["platelet"].max())
min_value = (pbc["platelet"].min())
print(max_value - min_value)

pbc.groupby("stage")["platelet"].agg(["mean"])
```

## Visualization

**I chose the variable platelet as a point of analysis and transformation. In particular, if it connects to the progression of stages of the Biliary Cirrhosis disease.**


```python
import matplotlib.pyplot as plt

pbc_no_na = pbc.dropna(subset=["stage"])

pbc_no_na.boxplot(column="platelet", by="stage")
plt.xlabel("Disease Stage")
plt.ylabel("Platelet Count (cubic mL/1000)")
plt.title("Platelet Count at Different Stages of Primary Biliary Cirrhosis")
plt.show()
```

## Ends with a short written interpretation of the result

Platelet ranges from 62 to 721 count. The box plot shows a decline in platelet counts from stage 1 to 4. Stage 1 had a median of 270, with moderate spread, compared with Stage 4, which had a median of 220. But the stage 4 does have several outliers. Additionally, the box plots overlap, suggesting that platelet count alone cannot indicate the patient's disease stage. But the principle can be applied to other variables to observe the impacts of D-penicillamine in primary biliary cirrhosis.
